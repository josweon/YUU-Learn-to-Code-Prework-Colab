## Part 2: New Information – Data Types, Variables, and Boolean Expressions

### 1. Data (or Values)
* Computer programs typically manipulate *data* (or "values")
  * Data can be numbers, names, or any text (and other things we'll investigate later)
* It's useful to think of computer programs as *accepting some input* and *producing some output*
  * The data (or values) are provided as input to the program, and some other data are produced as output

### 2. Basic Data Types

Most programming languages sort values into types. Three of the basic ones show up nearly everywhere:

* **int** (integers) values are whole numbers which have no fractional part
  * e.g., `42`, `-1`, `2026`
* **float** (short for "floating point") values are numbers which have a decimal point and possibly digits after the decimal point
  * e.g., `3.1415`, `212.`, `-1.5`
* **string** values are sequences of text surrounded by single or double quotes
  * e.g., `'I love programming!'`, `"Hello, world!"`

The names vary a little from language to language (you'll see `str` used as the short form for string in this one), but the categories are the same wherever you go.

We can use the builtin function `type()` to tell us the type of a value.

````python
# NOTE: always think about the response you expect before running a cell...
type('42')
````

````python
# Notice that you can chain values together with a comma, and the
# result will be a comma-separated list of results in parentheses...
type(-1), type(212.), type('hello')
````

### 3. Variables

* Named "boxes" inside the computer's memory into which you can put values (data)
* We can put a value into a variable by using an *assignment statement*, e.g.,
  * `year = 2026`
  * `name = 'Grace Hopper'`
* An assignment statement is not a statement of equality (like we're used to from mathematics)
  * Instead, it's a directive to put whatever *value* is on the right-hand side of the `=` into the variable on the left hand side

````python
year = 2026
year
````

````python
name = 'Grace Hopper'
name
````

#### Some thoughts about variable names...
* They may be arbitrarily long
* They may contain both letters and numbers, but must begin with a letter
* Uppercase letters are allowed, but by convention we don't use them (programming languages are often *case sensitive*, in other words `counter` and `Counter` are different variables)
* You should choose meaningful names for your variables:
  * `counter` instead of `c`
  * `cost_per_ounce` instead of `cpo`
  * `first_name` instead of `fn` or even `fname`
* As you can see above, variable names can include underscores, use them to make your variable names clearer
  * ...but do not start a variable name with an underscore
* Other languages have different conventions for stitching words together (you'll meet `costPerOunce` soon enough), but "make the name say what it holds" is universal

#### Careful!
* It's important to understand the difference between
  * *evaluating an expression* ... and ...
  * *printing a value*

````python
# First, let's remember that an assignment statement tells the computer to DO something
# ...but nothing is printed
name = 'Margaret Hamilton'
````

````python
name # ask the computer to evaluate this "expression", i.e.,
     # ... "Hey computer, show me the value inside this variable"
````

* When the notebook displays the value of an expression, it uses the same format you would use to enter its value, so in the case of strings, that means that it includes the quotes
* But when you call the `print()` function, the output displays the contents of the string without the quotes...

````python
print('Hello, my name is', name) # would the quotes make sense here?
````

### 4. Boolean Values and Boolean Expressions

* We've learned about `int`, `float`, and `str` data types
* Boolean values (`bool`) are the final basic data type
  * named after British mathematician [George Boole](https://en.wikipedia.org/wiki/George_Boole)
* A Boolean value can be one of two things: `True` or `False`
  * you can think of `True` as **yes** or **on**
  * ...and `False` is analogous to **no** or **off**
* At this point it might seem strange to have a variable which can only contain `True` or `False`, so in order to help us understand these kind of variables, we're first going to introduce the idea of a **Boolean expression**...
  * It's an expression that can only be `True` or `False`
  * Think of it as an **answer** to a **yes or no** question
  * Let's try it...as usual, do not run a cell until you have thought about what you expect the output of the cell to be

````python
4 > 3 # is this True or False?
````

````python
# the == operator is used to compare objects
# i.e., are the objects on either side of the == the same?
'Tailor' == 'tailor'
````

````python
3 * 9 == 27 # True or False?
````

The `!=` operator asks the opposite question: are these two things **different**?

````python
print(10 != 10)
print('Tailor' != 'tailor')
````

Let's try some Boolean expressions with variables in the mix...

````python
name == 'Margaret Hamilton'
````

````python
year > 2000 # 21st century? True or False?
````

A Boolean value can be stored in a variable, just like a number or a string:

````python
is_21st_century = year > 2000
print(is_21st_century)
print(type(is_21st_century))
````

### 5. From Algorithm to Code

In the first session, we said coding was converting a problem you want to solve into code which solves that problem. Now let's talk about how we will actually do it.

1. Write an algorithm, this should be understandable by another human (a non-programmer)
   * therefore, we will not include any programming-specific language constructs at this stage
   * once you have something that a non-programmer could read and perform the task, you're done with this step
2. Convert the algorithm to **pseudocode**
   * a mix of English and programming constructs
   * should include the structure, i.e., the indentation
   * ...but don't worry about the details/quirks of any specific programming language
   * think of this as a "stepping stone" to a code solution
   * the goal is to have steps which can be converted, one by one, into code
3. Convert each pseudocode step into actual code, and comment each line of code with the step it corresponds to
4. Let's see an example...

#### Worked example: convert a Fahrenheit temperature to Celsius

**Step 1: the algorithm** (a non-programmer could follow this)

```
1. Start with a temperature in Fahrenheit
2. Subtract 32 from it
3. Multiply that result by 5/9
4. Report the answer in Celsius
```

**Step 2: the pseudocode** (structure appears, language details don't)

```
set fahrenheit to 72
set celsius to (fahrenheit - 32) times 5/9
display celsius
```

**Step 3: the code.** Each line is commented with the pseudocode step it came from:

````python
fahrenheit = 72                          # set fahrenheit to 72
celsius = (fahrenheit - 32) * 5 / 9      # set celsius to (fahrenheit - 32) times 5/9
print('Celsius:', celsius)               # display celsius
````

* Notice how little the third step actually required
  * Most of the thinking happened in steps 1 and 2, where mistakes are cheap to fix
  * By the time you're writing code, you're mostly translating
* This is the process we'll use for the rest of the course, and it works in any language

The algorithm from Session 1:

````python
# Exercise 5, Step 1 - Algorithm (plain English, one step per line):
# 1. Ask for desired tip rate
# 2. Multiply the amount of the bill by that percentage
#
# Exercise 5, Step 2 - Pseudocode:
# 1. store the tip rate in a variable as floating point
# 2. compute the amount by multiplying tip rate by bill
# 3. print in dollar amount
# Exercise 5, Step 3 - Code (comment each line with its pseudocode step):
bill = 302

tip_rate = float(input("Enter tip rate as a percentage"))/100 # store the tip rate in a variable as floating point
tip_amount = bill * tip_rate # compute the amount by multiplying tip rate by bill

print(f"Tip amount = ${tip_amount:.2f}") # print in dollar amount

````
